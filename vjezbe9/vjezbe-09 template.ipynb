## Definicije
- k-particija skupa je rastav skupa $X=\{x^{(i)}\in \mathbb{R}^n: i=1,\cdots,m\}$ na $1\leq k \leq m$ disjunktnih, nepraznih podskupova $\pi_1, \cdots, \pi_k$
- broj svih k-particija skupa $X$ jednak je Stirlingovom broju druge vrste $|\mathcal{P}(X,k)| = \frac{1}{k!}\sum\limits_{j=1}^k (-1)^{(k-j)}{k \choose j}j^m$
- kvazimetrička funkcija: Funkcija $d:\mathbb{R}^n \times \mathbb{R}^n \rightarrow \mathbb{R}$ sa svojstvima
\begin{align}
d(x, y) \geq 0, \forall x, y \in \mathbb{R}^n \\
d(x, y) = 0 \iff x = y
\end{align}

- najbolji reprezentant (centar): Neka je $d:\mathbb{R}^n \times \mathbb{R}^n \rightarrow \mathbb{R}$ kvazimetrika, najbolji reprezentant skupa $\pi_j$ je 
$$
c_j^* \in \arg\min_{c \in \mathbb{R}^n}{\sum\limits_{i=1}^m d(c, x^{(i)})}
$$

U literaturi umjesto oznake $c^*$ za reprezentanta skupa možete pronaći oznaku $\mu^*$, što je zapravo ista stvar.

Kvazimetrike koje koristimo: 
- $d_{LS}(x,y) = ||x-y||^2_2$ (least square (LS) kvazimetrička funkcija)
- $d_1(x,y) = |x-y|$ ($\ell_1$ metrička funkcija, Manhattan metrika)

U slučaju $d_{LS}$ kvazimetrike, najbolji reprezentant je aritmetička sredina točaka (po komponentama u slučaju $A \subset \mathbb{R}^n$).

U slučaju $d_{1}$ kvazimetrike, najbolji reprezentant je medijan točaka (po komponentama u slučaju $A \subset \mathbb{R}^n$).

## k-means algoritam
### Kriterijska funkcija
- zadan je skup $X=\{x^{(i)} \in \mathbb{R}^n: i=1,\cdots,m\}$
- $k\geq 1$
- $\mathcal{F} : \mathcal{P}(X,k) \rightarrow \mathbb{R}_+$ (*suma udaljenosti do centra klastera po svim klasterima*)
- $\mathcal{F}(\Pi) = \sum\limits_{j=1}^{k}\sum\limits_{x^{(i)}\in\pi_j}d(c_j,x^{(i)})$

### Optimizacijski problem
- minimizirati po $ \Pi \in \mathcal{P}(X,k)$
- (globalno) optimalna particija je $\Pi ^* \in \arg\min\limits_{\Pi \in \mathcal{P}(X,k) }{\mathcal{F}(\Pi)}$ 

### Formiramo klaster: (princip minimalnih udaljenosti)
- $\pi_j = \{x^{(i)}: d(c_j, x^{(i)})\leq d(c_s,x^{(i)}),  \forall s=1,\cdots,k\} \; , j=1,\cdots,k$

### Pseudokod algoritma
1. odaberi početne centre $c_1, \cdots, c_k$  (*assignment step*)
2. po principu minimalnih udaljenosti odredi klastere $\pi_j = \{x^{(i)}: d(c_j, x^{(i)})\leq d(c_s,x^{(i)}), \forall s=1,\cdots,k\}\; , j=1,\cdots,k$ (*update step*)
3. definiramo nove centre  $c_j \in \arg\min \sum\limits_{x^{(i)}\in \pi_j}d(c_j, x^{(i)})$
4. GOTO 2 ako se novi centri razlikuju od prethodnih
5. RETURN $\pi_1, \dots, \pi_k$

### Calinski-Harabasz skor
Neka je $X \subset \mathbb{R}^n$ skup podataka, $\Pi$ neka particija, $(n_1, \dots, n_k)$ veličine klastera, $(c_1, \dots, c_k)$ reprezentanti klastera i $c$ reprezentant cijelog skupa $X$

CH skor se definira kao
$$
CH = \frac{\frac{1}{k-1}\sum_{j=1}^km_j||c_j - c||^2}{\frac{1}{m-k}\sum_{j=1}^k\sum_{a \in \pi_j}||c_j - a||^2}
$$
Veća vrijednost CH skora upućuje na međusobno dobro razdvojene klastere koji su "kompaktni".

## Meko klasteriranje
Za razliku od standardnog k-means algoritma, više ne pretpostavljamo da je neki podatak strogo pridružen samo jednom klasteru.

Kriterijsku funkciju smo ranije zadali kao
$$
\mathcal{F}(\Pi) = \sum\limits_{j=1}^{k}\sum\limits_{x^{(i)}\in\pi_j}d(c_j,x^{(i)})
$$
Ako za $i=1,\dots,m$ i $j=1,\dots,k$ uvedemo oznaku
$$
w_{ij} = \begin{cases} 1,  x^{(i)}\in\pi_j,\\ 0 \text{, inače} \end{cases}
$$
kriterijsku funkciju možemo definirati kao
$$
\mathcal{F}(\Pi) = \sum\limits_{i=1}^{m}\sum\limits_{j=1}^{k}w_{ij}d(c_j,x^{(i)})
$$
te vrijedi
$$
\begin{align}
\sum_{j=1}^k w_{ij} & = 1 \\
\sum\limits_{i=1}^{m}\sum\limits_{j=1}^{k}w_{ij} & = m
\end{align}
$$
Umjesto da zahtjevamo $w_{ij} \in \{0, 1\}$ kao ranije, zahtjevat ćemo samo da vrijede prethodna dva uvjeta i promatrati kriterijsku funkciju
$$
\Phi(c, W) = \sum\limits_{j=1}^k \sum\limits_{x^{(i)} \in \pi_j} w_{ij}d(c_j, x^{(i)}) 
$$

U kriterijskoj funkciji umjesto težina $w_{ij}$ se još često koristi $w_{ij}^q$, gdje je $q>1$ tzv. fuzzifier.

Postoje razni algoritmi za rješavanje problema mekog klasteriranja, a neke možete pronaći u udžbeniku [Klaster analiza i prepoznavanje geometrijskih objekata](http://www.mathos.unios.hr/images/uploads/857.pdf), poglavlje 7.

Jedan od algoritama je sljedeći:
1. Odaberi proizvoljan $q > 1$ (fuzzifier)
2. Slučajnim odabirom inicijaliziraj matricu pripadnosti $W \in \mathbb{R}^{m \times k}$
3. Izračunaj nove centre kao
$$
c_j = \frac{\sum_{i=1}^mw_{ij}^qx^{(i)}}{\sum_{i=1}^mw_{ij}^q}
$$
4. Izračunaj nove vrijednost matrice pripadnosti kao
$$
w_{ij} = \frac{1}{\sum_{l=1}^k\left(\frac{d(c_j, x^{(i)})}{d(c_l, x^{(i)})}\right)^{\frac{1}{m}}}
$$
5. Ako je norma razlike prethodnog i novog $W$ manja od nekog $\epsilon$, RETURN
6. GOTO 3

Glavni nedostatak standardnog k-means algoritma je osjetljivost na geometriju skupa.
